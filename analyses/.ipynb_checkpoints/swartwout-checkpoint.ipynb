```python
data_aggregate = {}
for k, v in data.items():
    data_aggregate[k] = np.array([0.0 for c in columns])
    for j in v:
        data_aggregate[k] += np.array(j[1]) / 100 * j[0]

df = pd.DataFrame.from_dict(data_aggregate,orient='index',columns=columns).round()
df = df.drop(["all"])

df
```
